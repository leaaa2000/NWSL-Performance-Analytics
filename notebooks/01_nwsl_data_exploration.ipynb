```python

import pandas as pd

print("Welcome to NWSL Performance Analytics!")
print("Pandas version:", pd.__version__)
```

```python
player_url = "https://raw.githubusercontent.com/leaaa2000/NWSL-Performance-Analytics/main/data/raw/adv_player_stats.csv"

team_url = "https://raw.githubusercontent.com/leaaa2000/NWSL-Performance-Analytics/main/data/raw/adv_team_stats.csv"

players = pd.read_csv(player_url)
teams = pd.read_csv(team_url)

print("Player dataset shape:", players.shape)
print("Team dataset shape:", teams.shape)
```

```python
# Take a look at both datasets
print("PLAYER DATASET:")
display(players.head())

print("TEAM DATASET:")
display(teams.head())
```

```python

# Get the season from the game ID
players["season"] = players["game_id"].str.extract(r"(\d{4})-\d{2}-\d{2}$")
teams["season"] = teams["game_id"].str.extract(r"(\d{4})-\d{2}-\d{2}$")

# See how many records are in each season
print("PLAYER DATA BY SEASON:")
print(players["season"].value_counts(dropna=False).sort_index())

print("\nTEAM DATA BY SEASON:")
print(teams["season"].value_counts(dropna=False).sort_index())
```

```python
# Check which columns have missing data
player_missing = players.isnull().sum()
team_missing = teams.isnull().sum()

print("PLAYER DATA - TOP 15 COLUMNS WITH MISSING VALUES:")
print(player_missing.sort_values(ascending=False).head(15))

print("\nTEAM DATA - TOP 15 COLUMNS WITH MISSING VALUES:")
print(team_missing.sort_values(ascending=False).head(15))
```

```python
# Keep the player stats I want to analyze
player_columns = [
    "game_id",
    "season",
    "team_id",
    "player_id",
    "first_name",
    "last_name",
    "position",
    "goals",
    "goal_assist",
    "total_shots",
    "total_pass",
    "accurate_pass"
]

players_clean = players[player_columns].copy()

# Check the smaller dataset
print("SELECTED PLAYER DATA:")
print("Shape:", players_clean.shape)

print("\nMISSING VALUES:")
print(players_clean.isnull().sum())

display(players_clean.head())
```

```python
# See which players are missing first names
missing_names = players_clean[
    players_clean["first_name"].isna()
]

print("Records with missing first names:", len(missing_names))

print("\nMissing first names by season:")
print(missing_names["season"].value_counts().sort_index())

print("\nExamples of missing first names:")
display(
    missing_names[
        ["season", "team_id", "player_id", "first_name", "last_name"]
    ].head(15)
)
```

```python

# Create a display name without losing player records

players_clean["player_name"] = (
    players_clean["first_name"]
    .fillna("")
    .str.strip()
    + " "
    + players_clean["last_name"].fillna("").str.strip()
).str.strip()

# Check the results
print("Missing player display names:")
print(players_clean["player_name"].isna().sum())

print("\nExamples of players with missing first names:")
display(
    players_clean.loc[
        players_clean["first_name"].isna(),
        ["first_name", "last_name", "player_name"]
    ].drop_duplicates().head(15)
)
```

```python

# Check for completely identical rows
exact_duplicates = players_clean.duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)

# Check for duplicate player-match records
player_match_duplicates = players_clean.duplicated(
    subset=["game_id", "team_id", "player_id"]
).sum()

print("Duplicate player-match records:", player_match_duplicates)

# Check for blank player display names
blank_names = (
    players_clean["player_name"].fillna("").str.strip() == ""
).sum()

print("Blank player display names:", blank_names)
```

```python
players_clean.to_csv(
    "nwsl_players_cleaned.csv",
    index=False
)

print("Cleaned player dataset saved successfully!")
print("Rows:", players_clean.shape[0])
print("Columns:", players_clean.shape[1])
```

```python
# Look at the team data and its columns
print("TEAM DATASET SHAPE:")
print(teams.shape)

print("\nALL TEAM COLUMNS:")
print(teams.columns.tolist())

print("\nFIRST FIVE ROWS:")
display(teams.head())
```

```python

# Pick the team stats I want to compare
team_columns = [
    "game_id",
    "season",
    "status",
    "team",
    "goals",
    "goal_assist",
    "total_shots",
    "total_pass",
    "accurate_pass"
]

# Make sure the columns are in the dataset
available = [col for col in team_columns if col in teams.columns]
missing = [col for col in team_columns if col not in teams.columns]

print("Available columns:", available)
print("\nMissing columns:", missing)

teams_clean = teams[available].copy()

# Check the selected team data
print("\nCleaned team dataset shape:")
print(teams_clean.shape)

print("\nMissing values:")
print(teams_clean.isnull().sum())

display(teams_clean.head())
```

```python
# Check for duplicate team records
exact_duplicates = teams_clean.duplicated().sum()

team_match_duplicates = teams_clean.duplicated(
    subset=["game_id", "team"]
).sum()

print("Exact duplicate rows:", exact_duplicates)
print("Duplicate team-match records:", team_match_duplicates)

# Make sure the stats are numbers
stats = [
    "goals",
    "goal_assist",
    "total_shots",
    "total_pass",
    "accurate_pass"
]

for col in stats:
    teams_clean[col] = pd.to_numeric(
        teams_clean[col], errors="coerce"
    )

# Look for missing or incorrect values
print("\nInvalid or missing numeric values:")
print(teams_clean[stats].isna().sum())

print("\nNegative values:")
print((teams_clean[stats] < 0).sum())

# Completed passes shouldn't be higher than total passes
invalid_passes = (
    teams_clean["accurate_pass"] >
    teams_clean["total_pass"]
).sum()

print("\nMatches with completed passes exceeding total passes:",
      invalid_passes)
```

```python
teams_clean.to_csv(
    "nwsl_teams_cleaned.csv",
    index=False
)

print("Cleaned team dataset saved successfully!")
print("Rows:", teams_clean.shape[0])
print("Columns:", teams_clean.shape[1])
```
